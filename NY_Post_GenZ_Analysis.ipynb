### St. John's University × New York Post — Project Hamilton

## 6. Predictive Analysis

Beyond descriptive patterns, predictive models were explored to determine whether respondent characteristics and news-consumption behavior could help explain differences in preferred news formats and content interests.

### Predicting Preference for Short-Form Video

Because short-form video emerged as the most preferred news format, a binary classification model was developed to examine whether respondent characteristics and consumption behavior were associated with choosing short-form video over other formats.

The target variable was coded as:

- **1:** Short-form video preferred
- **0:** Any other news format preferred

Predictors included age, gender, news-consumption frequency, and time of consumption.

```python
model_df = analysis_df[
    [
        "preferred_consumption",
        "age_clean",
        "gender_clean",
        "consumption_freq_clean",
        "time_of_consumption"
    ]
].copy()

# Remove observations missing any modeling variable
model_df = model_df.dropna()

# Binary target: short-form video vs. other formats
model_df["short_form_preference"] = (
    model_df["preferred_consumption"]
    == "Short-form videos (TikTok, Instagram Reels, YouTube Shorts)"
).astype(int)

print("Modeling observations:", len(model_df))
print("\nTarget distribution:")
print(model_df["short_form_preference"].value_counts())

print("\nTarget percentages:")
print(
    model_df["short_form_preference"]
    .value_counts(normalize=True)
    .mul(100)
    .round(1)
)
```

```python
def convert_age(age):
    if age == "18–20":
        return 19
    elif age == "21–23":
        return 22
    elif age == "24–26":
        return 25
    else:
        try:
            return float(age)
        except:
            return np.nan

model_df["age_numeric"] = model_df["age_clean"].apply(convert_age)

model_df[["age_clean", "age_numeric"]].drop_duplicates().sort_values("age_numeric")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

X = model_df[
    [
        "age_numeric",
        "gender_clean",
        "consumption_freq_clean",
        "time_of_consumption"
    ]
]

y = model_df["short_form_preference"]

categorical_features = [
    "gender_clean",
    "consumption_freq_clean",
    "time_of_consumption"
]

numeric_features = ["age_numeric"]

preprocessor = ColumnTransformer(
    transformers=[
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features),
        ("num", StandardScaler(), numeric_features)
    ]
)

model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("classifier", LogisticRegression(max_iter=1000))
    ]
)

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

model.fit(X_train, y_train)

y_pred = model.predict(X_test)

print("Training observations:", len(X_train))
print("Testing observations:", len(X_test))
print(f"Accuracy: {accuracy_score(y_test, y_pred):.3f}")

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
# Compare model accuracy with a majority-class baseline

baseline_accuracy = y_test.value_counts(normalize=True).max()
model_accuracy = accuracy_score(y_test, y_pred)

print(f"Baseline accuracy: {baseline_accuracy:.3f}")
print(f"Logistic regression accuracy: {model_accuracy:.3f}")
print(f"Improvement over baseline: {(model_accuracy - baseline_accuracy):.3f}")
```

### Model Interpretation

The logistic regression achieved approximately **56.5% accuracy** on the held-out test set, compared with a majority-class baseline of approximately **55.6%**.

The model therefore provided little improvement over simply predicting the most common class. It also identified only about **21% of respondents who actually preferred short-form video**, indicating weak predictive performance for the positive class.

This suggests that age, gender, news-consumption frequency, and time of consumption alone are not sufficient to explain an individual's preference for short-form video.

Rather than treating demographic characteristics as strong predictors of format preference, the results suggest that additional behavioral variables—such as platform usage, engagement style, and existing social-media habits—may provide more useful explanatory information.

### Model 2: Adding Behavioral Variables

Because demographic and basic consumption variables provided little predictive value, a second model was developed incorporating respondents' actual media behavior.

Additional predictors included Instagram, TikTok and YouTube usage, engagement style, and preference for receiving news through social media. The objective was to test whether behavioral characteristics provide more predictive information than demographics alone.

```python
behavior_df = analysis_df[
    [
        "preferred_consumption",
        "age_clean",
        "gender_clean",
        "consumption_freq_clean",
        "time_of_consumption",
        "engagement_type",
        "socialmediastyle",
        "Uses (Instagram)",
        "Uses (TikTok)",
        "Uses (YouTube)"
    ]
].copy()

behavior_df = behavior_df.dropna()

# Create target
behavior_df["short_form_preference"] = (
    behavior_df["preferred_consumption"]
    == "Short-form videos (TikTok, Instagram Reels, YouTube Shorts)"
).astype(int)

# Convert age
behavior_df["age_numeric"] = behavior_df["age_clean"].apply(convert_age)

behavior_df = behavior_df.dropna(subset=["age_numeric"])
```

```python
X2 = behavior_df[
    [
        "age_numeric",
        "gender_clean",
        "consumption_freq_clean",
        "time_of_consumption",
        "engagement_type",
        "socialmediastyle",
        "Uses (Instagram)",
        "Uses (TikTok)",
        "Uses (YouTube)"
    ]
]

y2 = behavior_df["short_form_preference"]

categorical_features2 = [
    "gender_clean",
    "consumption_freq_clean",
    "time_of_consumption",
    "engagement_type",
    "socialmediastyle"
]

numeric_features2 = [
    "age_numeric",
    "Uses (Instagram)",
    "Uses (TikTok)",
    "Uses (YouTube)"
]

preprocessor2 = ColumnTransformer(
    transformers=[
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features2),
        ("num", StandardScaler(), numeric_features2)
    ]
)

model2 = Pipeline(
    steps=[
        ("preprocessor", preprocessor2),
        ("classifier", LogisticRegression(max_iter=1000))
    ]
)

X2_train, X2_test, y2_train, y2_test = train_test_split(
    X2,
    y2,
    test_size=0.20,
    random_state=42,
    stratify=y2
)

model2.fit(X2_train, y2_train)

y2_pred = model2.predict(X2_test)

baseline2 = y2_test.value_counts(normalize=True).max()
accuracy2 = accuracy_score(y2_test, y2_pred)

print(f"Baseline accuracy: {baseline2:.3f}")
print(f"Model 2 accuracy: {accuracy2:.3f}")
print(f"Improvement over baseline: {accuracy2 - baseline2:.3f}")

print("\nConfusion Matrix:")
print(confusion_matrix(y2_test, y2_pred))

print("\nClassification Report:")
print(classification_report(y2_test, y2_pred))
```

### Model 2 Interpretation

Adding behavioral variables improved predictive performance. Model 2 achieved **65.4% test accuracy**, compared with a majority-class baseline of **56.1%**, representing an improvement of approximately **9.3 percentage points**.

The model also identified **45% of respondents who preferred short-form video**, compared with only 21% in the demographic-focused model.

These results suggest that respondents' digital behaviors and engagement patterns contain more useful information for predicting short-form video preference than demographic characteristics and basic consumption patterns alone.

However, the model's performance remains moderate rather than highly predictive. The findings should therefore be interpreted as evidence of useful behavioral signals rather than as a production-ready prediction system.

### Model Insights

The modeling results indicate that behavioral information improved prediction of short-form video preference compared with demographic and basic consumption characteristics alone.

Model 1, which relied primarily on demographics and general consumption patterns, performed only marginally better than the majority-class baseline. After adding platform usage and engagement behavior, Model 2 improved test accuracy to **65.4%**, approximately **9.3 percentage points above its baseline**.

Among the behavioral variables, **TikTok usage showed a positive association with short-form video preference** (odds ratio ≈ 1.34). This is consistent with the broader descriptive findings showing strong adoption of TikTok and substantial demand for short-form video content.

## 7. Business Recommendations

The analysis points to several opportunities for the New York Post to strengthen engagement with the surveyed Gen Z audience.

### 1. Make short-form video a core discovery format
Short-form video was the most preferred news format (43.8%) and the most requested app feature (55.3%). Instagram and TikTok were also the two most widely used platforms among respondents.

**Recommendation:** Develop concise video recaps designed for Instagram Reels, TikTok, YouTube Shorts, and the New York Post's own digital experience, with clear pathways to deeper coverage.

### 2. Compete with social media on convenience
Nearly 59.3% of respondents said social media already covers their news needs, while 39.0% cited time or format constraints as barriers to traditional news consumption.

**Recommendation:** Treat social platforms as an important discovery channel and make the transition from social content to full New York Post coverage fast, mobile-friendly, and low-friction.

### 3. Prioritize personalization
Topic and local personalization was requested by 50.2% of respondents, making it the second-most desired app feature.

**Recommendation:** Allow users to personalize their experience around topics, local news, sports, entertainment, and other interests rather than relying on a single universal news feed.

### 4. Reduce barriers to continued engagement
Cost, paywalls, and advertising were barriers for 33.2% of respondents, while 35.5% selected student pricing as a desirable app feature.

**Recommendation:** Explore student-focused pricing and lower-friction introductory experiences while giving users greater control over notifications and content discovery.

## 8. Conclusion

This analysis suggests that engaging the surveyed Gen Z audience requires more than moving traditional news content onto digital platforms. Respondents already encounter much of their news through social media and show a strong preference for short, accessible, and personalized formats.

The predictive analysis reinforced this distinction. Demographic and general consumption characteristics provided little predictive value for short-form video preference, while adding behavioral and platform-usage variables improved model performance.

For the New York Post, the findings point toward a strategy centered on **short-form content, social discovery, personalization, and reduced user friction**, while using these experiences as pathways to deeper journalism and continued engagement.

### Limitations

The survey primarily represents students from St. John's University and should not be treated as representative of the entire Gen Z population. Survey responses are self-reported, and the predictive models demonstrate associations rather than causal relationships. Future research could validate these findings using a broader sample and behavioral engagement data.
